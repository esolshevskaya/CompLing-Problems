```python
import re

NAMES = [
    "Иванов И. И.",
    "И. И. Иванов",
    "Иванов И.И.",
    "А.С. Пушкин",
    "Пушкин А.С.",
    "Д. М. Шеберле",
    "М. Кюри",
    "Ж.К. Ван Дамм",
]

surname = r'[А-ЯЁ][а-яё]+(?:\s+[А-ЯЁ][а-яё]+)?'

pattern = re.compile(
    rf'(?:(?P<s>{surname})\s*(?P<i1>[А-ЯЁ])\.\s*(?P<i2>[А-ЯЁ])?\.|'
    rf'(?P<i3>[А-ЯЁ])\.\s*(?P<i4>[А-ЯЁ])?\.?\s*(?P<t>{surname}))'
)

for name in NAMES:
    m = pattern.fullmatch(name)
    g = m.groupdict()
    s = g['s'] or g['t']
    i1 = g['i1'] or g['i3']
    i2 = g['i2'] or g['i4']
    print(f"{s} {i1}." + (f"{i2}." if i2 else ""))
```

Но структурное сходство всё равно остаётся: в обоих случаях сильная исходная рамка может сделать пересмотр этой рамки менее вероятным именно тогда, когда он особенно необходим.

В статье есть ещё один важный для меня момент: реакция модели сильно зависела от того, насколько близко к потенциально опасному действию находилось явное указание остановиться. Когда инструкция о границах взаимодействия появлялась непосредственно перед действием, Mythos 5 останавливался примерно в 90% случаев, а если та же инструкция появлялась за три шага до этого, показатель снижался примерно до 40%. Это показывает, что даже правильное понимание ограничения не обязательно сохраняется как достаточное основание для остановки. Текущий контекст и уже выбранное направление действий могут оказаться сильнее более раннего предупреждения. В человеческом мышлении мне тоже кажется возможным похожее явление, когда после принятия решения человек начинает воспринимать последующие события прежде всего как аргументы за продолжение уже выбранного курса.

В итоге статья скорее подняла вопрос, чем дала готовый ответ. Если система способна интерпретировать противоречащие данные в пользу уже выбранной стратегии, одной способности распознавать опасные ситуации недостаточно — важно умение вовремя остановиться и допустить, что сама исходная интерпретация была ошибочной. Вероятно, это и есть одна из самых интересных проблем, показанных в статье. И, возможно, самое опасное здесь не то, что модель может ошибиться, а то, что она способна оказаться достаточно последовательной в оправдании собственной ошибки.
